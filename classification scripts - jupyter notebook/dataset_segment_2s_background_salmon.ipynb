```python
import os
from pathlib import Path

import librosa
import soundfile as sf
from tqdm import tqdm

# =====================================================
# PATHS
# =====================================================

SOURCE_DIR = "/home/feliciano/feeding_dataset/Class_0_Background"

OUTPUT_DIR = (
    "/media/feliciano/Aux/AI_AFS_DATASET/"
    "behavior_dataset/Background"
)

# =====================================================
# SETTINGS
# =====================================================

TARGET_SEGMENTS = 1500
SEGMENT_DURATION = 2  # seconds
TARGET_SR = 16000

# =====================================================
# CREATE OUTPUT DIRECTORY
# =====================================================

os.makedirs(OUTPUT_DIR, exist_ok=True)

# =====================================================
# FIND WAV FILES
# =====================================================

wav_files = sorted(Path(SOURCE_DIR).rglob("*.wav"))

print(f"Found {len(wav_files)} WAV files")

segment_count = 0

# =====================================================
# PROCESS FILES
# =====================================================

for wav_file in tqdm(wav_files, desc="Processing files"):

    if segment_count >= TARGET_SEGMENTS:
        break

    try:

        audio, sr = librosa.load(
            wav_file,
            sr=TARGET_SR,
            mono=True
        )

        samples_per_segment = TARGET_SR * SEGMENT_DURATION

        total_segments = len(audio) // samples_per_segment

        for i in range(total_segments):

            if segment_count >= TARGET_SEGMENTS:
                break

            start = i * samples_per_segment
            end = start + samples_per_segment

            segment = audio[start:end]

            output_file = os.path.join(
                OUTPUT_DIR,
                f"background_{segment_count:05d}.wav"
            )

            sf.write(
                output_file,
                segment,
                TARGET_SR
            )

            segment_count += 1

    except Exception as e:
        print(f"Error with {wav_file}")
        print(e)

print("\nDone.")
print(f"Created {segment_count} background segments")
print(f"Saved to: {OUTPUT_DIR}")
```
